## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
from IPython.display import Markdown, display

display(Markdown("""
### Why ML Beats a Fixed Rule

A fixed rule such as "refresh pages with CTR below 2%" ignores many important factors.

Machine learning can combine multiple signals including CTR, impressions,
ranking position, and content age to estimate which pages have the highest opportunity.

This produces a more accurate prioritization than using a single threshold.
"""))
```
