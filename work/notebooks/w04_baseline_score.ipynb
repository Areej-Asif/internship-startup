## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import os

# Create normalized scores
df["age_score"] = (
    df["content_age_days"] / df["content_age_days"].max()
)

df["impression_score"] = (
    df["impressions_90d"] / df["impressions_90d"].max()
)

df["ctr_opportunity"] = (
    1 - (df["ctr"] / df["ctr"].max())
)

# Create baseline action score
df["action_score"] = (
    df["age_score"] * 0.4
    +
    df["impression_score"] * 0.3
    +
    df["ctr_opportunity"] * 0.3
)


# Reason codes
def get_reason(row):
    if row["age_score"] > 0.6:
        return "STALE_CONTENT"
    elif row["ctr_opportunity"] > 0.6:
        return "CTR_OPPORTUNITY"
    else:
        return "HIGH_VALUE_REFRESH"


df["reason_code"] = df.apply(get_reason, axis=1)

# Action label
df["action_label"] = "REFRESH_CONTENT"


# Rank queue
queue = df.sort_values(
    by="action_score",
    ascending=False
)


# Save CSV
os.makedirs(
    "flyrank-ml-internship-starter/work/outputs",
    exist_ok=True
)

queue.to_csv(
    "flyrank-ml-internship-starter/work/outputs/baseline_action_score.csv",
    index=False
)


print("✅ baseline_action_score.csv created")
print("Rows:", len(queue))

queue.head(10)
```
